### Q1 · One quantitative variable
**How many hours per week do Northgate students work?** Describe what's typical, how spread out it is, and whether anyone stands apart.
→ *histogram* (add a *box plot* if it helps). Column: `work_hours_per_week` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)
ax.set_title("Work Hours for Northgate students")
ax.set_xlabel("Time")
ax.set_ylabel("Number of students")
plt.show()
```

### Q2 · One categorical variable
**How does enrollment compare across majors?**
→ *sorted bar chart*. Column: `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
order = students["major"].value_counts().index          # sort by count
fig, ax = plt.subplots()
sns.countplot(data=students, y="major", order=order, color=NAVY, ax=ax)
ax.set_title("Number of students by major")
ax.set_xlabel("Number of students")
ax.set_ylabel("Major")
plt.show()
```

**STAR story — Q2** *(write in your own words)*

- **Scope —** Counted total enrollments across every program. I used the major field from student_records.csv.
- **Track —** Compares department sizes. The goal was to find Northgate's biggest and smallest major groups.

- **Articulate —** Cybersecurity is the top group, with more than 500 students. Data Science and Information Science come next. Information Technology is also in that range, at about 280 students each.
- **Respond —** Some majors show up twice because of misspellings. For example, "Cyber Security" differs from "Cybersecurity." Also, "Info Technology" differs from "Information Technology." Fixing these labels should give the Provost a correct student count for budgeting.

### Q3 · A quantity, split by category
**Does final GPA differ across majors?** Read the exceptions, not just the boxes.
→ *box plot per major*. Columns: `final_gpa` and `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=students, x="final_gpa", y="major", color=NAVY, ax=ax)
ax.set_title("Final GPA by major — boxes agree, outliers don't")
ax.set_xlabel("Final GPA")
ax.set_ylabel("Major")
plt.show()
```

### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Final GPA vs. commute miles")
ax.set_xlabel("Commute Distance (Miles)")
ax.set_ylabel("Final GPA (0.0 - 4.0 Scale)")
plt.show()
```

**STAR story — Q4** *(write in your own words)*

- **Scope —** It shows how commute_miles links to final_gpa for all students in student_records.csv.

- **Track —** Checks if students who drive farther tend to do worse in school.
- **Articulate —** Most students are under 20 miles. The line from the regression still moves down. That means the average GPA falls for those who commute more than 30 miles.
- **Respond —** Longer trips seem to pull down GPA results. After that, I would see whether those long-commute students spend less time on LMS tasks and whether they often skip morning classes.

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             color=NAVY, marker="o", ax=ax)
ax.set_title("Average active minutes per student, by week")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Avg. minutes active")
plt.show()
```

**STAR story — Q5** *(write in your own words)*

- **Scope —** It measures minutes_active, the average weekly active minutes, for each week (week) across all term weeks. And I used the weekly_activity.csv.
- **Track —** This checks how a student’s LMS use changes over the semester.
- **Articulate —** For most of the term, the weekly active time stays steady near 210 minutes. Then in Week 9 it falls to about 90 minutes. After that, it jumps back up in Week 10.
- **Respond —** The Week 9 drop fits with Spring Break. Next, I would review student level activity paths to see whether some students slow down even when there is no planned university break.
